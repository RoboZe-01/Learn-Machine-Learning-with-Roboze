# Model Tuning 

- It is Nothing but selecting the best model parameters for the model to get optimal results 

## Core Concept : Model Parameters vs hyperparameters


 **Model Parameters** : 
- These are the internal parameters that model sets to learn from the data 
- Key idea is "We don't set them , these are already set by the model"

**Hyperparameters** : 
- These are the external parameters we set to get the best model output as possible 
- They control the learning process itself
- Ml practitioner has to set them before the training itself 



But if we randomly set them , models may get underfit or overfit 

so the goal for the tuning is :
1. to find the best combination of hyperparameters 
2. so that model performs better on new , unseen data - not just training data

In short - tuning helps us squeeze out best possible performance from the model
