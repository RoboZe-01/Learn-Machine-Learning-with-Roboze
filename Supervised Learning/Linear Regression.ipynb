## Core idea : 
Find the **Best-fit straight line** through the data points

## Cost Function :

- Quantifies the error between a models prediction and actual value
- Cost function is the summation of all the resuadal error 
- Difference between predicted and value of the line 
- There are multiple cost function for multiple line 
- You have to minimize the cost function by changing the values of $\theta_0$ , and theta $\theta_1$ 



## Gradient Descent

- It is the a optimization algorithm in machine learning that iteratively adjusts a model's internal parameters ( weights/biases) to minimize a cose (or loss) function 

- we Adjust the values of $\theta_0$ and $\theta_1$ and check the various lines through data points then select the best fit line when we get the lowest cost function

- The lowest value of the cost function on the gradient descent curve is global minima point 
( so it the process to find the global minima point )

### Process :   
Data points --> assuming th values of $\theta_1$ and $\theta_0$ --> find values of cost function for each values --> plot the graph --> select the lowest point on the graph ( global minima) --> we will get best fit line

## Learning rate 

- In simple words step size

- It is a crucial hyperparameter that controls the step size during model training 

- It determines how much to adjust parameters ( weights / biases ) with each update to minimize the error

- We prefer slow learning rate to so the values vary in lower rate so the accuracy increases

## Hyperline 

- It is a decision boundary that seperates data points in a feature space , acting as a generalization of line ( in 2d) or a plane ( in 3d) to higher dimensions , with one less dimension than the space itself .
