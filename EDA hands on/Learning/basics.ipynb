# Types of EDA analysis

- Number of features to be analysed is **one** --> **Univariate analysis**
- Number of features to be analysed is **Two** --> **Bivariate**
- Number of features to be analysed is **More than two** --> **Multivariate**


## Univariate analysis
( to understand the distribution )

**1. Numerical variables**: 
- Summary stats : Count , Mean , Median , std  , min , max , quartiles  , IQR , skewness
- Visuals : histograms , boxplot , density plot
- check for multimodality ( multiple peaks ) and heavy tails 

**2. Categorical variables**: 
- frequency counts
- Bar charts and Pareto charts
- check rare categories ( low- frequency levels ) and cardinality( number of categories)

**3.Data/time variables:**
- check for time range , gaps , time granuality
- Visuals : Time series plot , seasonality checks , Heatmaps , bar chart  (month/year)

**4.Text variables** : 
- token counts , most common words , length distributions 
- Histograms of text lengths 
- word cloud




## Bivariate Analysis
( usually for relationships between features )

**1.Numeric - Numeric:**
- Scatter plots , correlation coefficient
- Regression line to visualise trend

**2. Numeric - Categorical:**
- boxplot of violin plot grouped by category
- compute group means / medians
- ANOVA for significant differences
**3. Categorical - Categorical:**
- Contingency tables , stacked bar charts
- Chi-square test for independence

**4. TIme series relationship:**
- Lag plots , autocorrelations
- rolling averages
