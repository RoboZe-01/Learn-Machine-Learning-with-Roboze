## Univariate analysis

### Numerical data

```python
df.columns
```

```python
num_cols= ['Pregnancies', 'Glucose', 'BloodPressure', 'SkinThickness', 'Insulin',
       'BMI', 'DiabetesPedigreeFunction', 'Age', 'Outcome']
```

### Statistical summary

```python
summary = df.describe().T
summary['median']= df.median
summary['IQR']=summary['75%']-summary['25%']
summary['skewness']= df.skew()
print("Numeric Summary Statistics:\n")
summary
```

```python
# histogram , boxplot for numerical columns 

for cols in num_cols:
    plt.figure(figsize=(12,4))

    plt.subplot(1,3,1)
    sns.histplot(df[cols] , kde=True)
    plt.title(f'Histogram of {cols}')
    
    plt.subplot(1,3,2)
    sns.boxplot(y=df[cols])
    plt.title(f'Boxplot of {cols}')
    
    plt.subplot(1,3,3)
    sns.kdeplot(df[cols],fill=True)
    plt.title(f'Density plot of {cols}')

    plt.tight_layout()
    plt.show()

```
